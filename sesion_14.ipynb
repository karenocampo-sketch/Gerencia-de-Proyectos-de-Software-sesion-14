Valor Ganado (EVM)
El Valor Ganado (EVM) es una técnica que permite saber cómo está funcionando un proyecto en términos de dinero y tiempo.
La idea principal es responder tres preguntas:
¿Dónde estamos ahora?
¿Cómo vamos a terminar?
¿Qué debemos hacer para cumplir el objetivo?

1. Diagnóstico: ¿dónde estamos hoy?
Aquí analizamos el estado actual del proyecto.
Los dos indicadores principales son:
CPI – Índice de rendimiento de costos
Nos dice qué tan bien estamos utilizando el dinero.
CPI = 1.0: estamos exactamente en lo presupuestado.
CPI > 1.0: estamos teniendo un buen rendimiento del dinero.
CPI < 1.0: estamos gastando más de lo que corresponde al trabajo realizado.
Por eso, el 1.0 es el punto de referencia.
SPI – Índice de rendimiento del cronograma
Nos dice qué tan bien estamos cumpliendo el cronograma.
SPI = 1.0: vamos según lo planeado.
SPI > 1.0: estamos adelantados.
SPI < 1.0: estamos atrasados.
Una forma fácil de recordarlo:
CPI = dinero
SPI = tiempo

2. Las tres curvas principales
Para entender el EVM necesitamos conocer tres valores:
PV – Valor Planificado: cuánto trabajo y presupuesto estaba previsto realizar hasta ese momento.
EV – Valor Ganado: cuánto valor tiene realmente el trabajo que ya se completó.
AC – Costo Real: cuánto dinero realmente hemos gastado.
La comparación entre estos tres valores permite detectar si el proyecto está gastando más, menos, adelantado o atrasado respecto al plan.

3. Bola de cristal: ¿cómo terminaremos?
Después de saber cómo estamos actualmente, podemos hacer un pronóstico.
EAC – Estimación al finalizar
Responde:
¿Cuánto esperamos que cueste todo el proyecto cuando termine?
ETC – Estimación para completar
Responde:
¿Cuánto dinero necesitamos todavía para terminar el proyecto?
VAC – Variación al finalizar
Compara el presupuesto original con lo que esperamos gastar finalmente.
En otras palabras:
EAC = cuánto costará todo
ETC = cuánto falta gastar
VAC = diferencia frente al presupuesto

4. TCPI: ¿qué necesitamos hacer de ahora en adelante?
El TCPI indica el nivel de rendimiento que necesitamos conseguir en el trabajo que queda para alcanzar nuestro objetivo.
Es decir, no pregunta:
“¿Cómo nos ha ido?”
Sino:
“¿Qué tan eficientes tenemos que ser desde este momento para cumplir la meta?”
Por eso puedes verlo como una meta de eficiencia para el equipo.

5. Dashboard ejecutivo
Toda esta información puede reunirse en un dashboard para que el equipo o los directivos puedan entender rápidamente el estado del proyecto.
El dashboard debería mostrar:
Las curvas PV, EV y AC.
CPI y SPI comparados con el valor ideal de 1.0.
Las variaciones actuales de costo y cronograma.
Las proyecciones EAC, ETC y VAC.
El TCPI, para saber qué rendimiento necesitamos mantener.
La idea es que el dashboard permita ver:
Pasado → presente → futuro
Lo que se planeó, lo que está pasando y cómo probablemente terminaremos.

Gestión de riesgos
Un riesgo es un evento incierto que puede afectar los objetivos del proyecto.
Puede tener un efecto:
Negativo: una amenaza.
Positivo: una oportunidad.
Por ejemplo:
“¿Cuánto dinero podría perder el proyecto si falla un API?”
Eso sería analizar el posible impacto de un riesgo.

1. Análisis cualitativo
Es la primera evaluación del riesgo.
Principalmente analizamos:
Probabilidad + Impacto
Por ejemplo, un riesgo puede tener:
Probabilidad alta.
Impacto alto.
Eso nos permite identificar cuáles riesgos necesitan mayor atención.

2. Análisis cuantitativo
Aquí vamos un paso más allá: intentamos medir económicamente los riesgos y sus posibles efectos.
Árbol de decisiones
Muestra diferentes caminos posibles y sus resultados.
Diagrama de Tornado
Permite identificar qué variables tienen mayor influencia sobre el resultado del proyecto.
Simulación de Monte Carlo
Simula muchos escenarios posibles para entender cómo diferentes riesgos e incertidumbres pueden afectar el proyecto.
La idea importante es:
No analiza solamente un futuro posible, sino muchos escenarios posibles.

VME – Valor Monetario Esperado
El VME permite estimar económicamente un riesgo teniendo en cuenta su probabilidad.
La lógica es:
VME = Probabilidad × Impacto económico
Ejemplo:
Si existe un 20 % de probabilidad de perder $1.000.000:
0,20 × $1.000.000 = $200.000
Ese sería el valor monetario esperado de esa posible pérdida.

Estrategias de respuesta a los riesgos
Una vez identificamos y analizamos los riesgos, debemos decidir cómo responder ante ellos.
Para eso utilizamos estrategias y mantenemos un registro de riesgos.
El registro permite documentar:
Qué riesgo existe.
Qué probabilidad tiene.
Qué impacto puede generar.
Qué respuesta se aplicará.
Cómo se hará seguimiento.

📊 Valor ganado y desempeño del proyecto
Misterio del desempeño:
Son las dudas que aparecen sobre cómo está avanzando el proyecto: ¿estamos haciendo lo que planeamos y gastando lo que habíamos presupuestado?
💰 Bases del valor ganado
El valor ganado sirve como un diagnóstico del proyecto, porque permite comparar lo que se planeó con lo que realmente está pasando.
Se analizan 3 dimensiones:
PV – Valor Planificado: cuánto trabajo se planeó hacer y cuánto presupuesto estaba autorizado para hacerlo.
AC – Costo Real: cuánto dinero realmente se gastó en el trabajo realizado.
EV – Valor Ganado: cuánto valor representa el trabajo que realmente se ha realizado.
👉 En palabras sencillas: PV = lo planeado, AC = lo gastado y EV = lo realizado.
🔎 Análisis de variaciones
Es como una prueba de diagnóstico para saber si el proyecto tiene problemas.
Permite identificar, por ejemplo:
Si se está gastando más o menos de lo planeado.
Si el proyecto está adelantado o atrasado en el cronograma.
Si existe una hemorragia financiera, es decir, un gasto excesivo que puede afectar el proyecto.
📋 Reporte de desempeño
El reporte de desempeño reúne y analiza la información del proyecto.
Sus principales funciones son:
Consolidar: reunir toda la información.
Calcular: obtener los indicadores y variaciones.
Interpretar: entender qué significan los resultados.
Asesorar: utilizar la información para apoyar la toma de decisiones.
🎯 Beneficio principal
El reporte de desempeño entrega información real sobre cómo va el proyecto, lo que ayuda a tomar decisiones difíciles y definir mejores estrategias.
En resumen: el valor ganado permite comparar lo que planeamos, lo que gastamos y lo que realmente hicimos, para saber si el proyecto va por buen camino o necesita ajustes.
